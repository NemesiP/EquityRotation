## Reading the map

- **Leading:** relative direction and momentum are above 100.
- **Weakening:** relative direction remains above 100 while momentum falls below it.
- **Lagging:** both measures are below 100.
- **Improving:** momentum is above 100 while relative direction remains below it.

The calculations are a transparent EMA-based RRG proxy, not the proprietary JdK formula. Use the playback controls to move both charts through history; the summary and table remain pinned to the latest observation.

## Optional Bloomberg source

Bloomberg identifiers are canonicalized without uppercasing their yellow key: `SPY US EQUITY` becomes `SPY US Equity`. To use a Terminal-backed client such as `xbbg`, pass a custom loader to `launch_notebook_dashboard`:

```python
from xbbg import blp
from rrg import prepare_bloomberg_prices

def bloomberg_loader(tickers, start, end):
    raw = blp.bdh(list(tickers), flds="PX_LAST", start_date=start, end_date=end)
    return prepare_bloomberg_prices(raw, tickers)

dashboard = launch_notebook_dashboard(
    auto_load=False,
    price_loader=bloomberg_loader,
    data_source_name="Bloomberg",
)
```

`prepare_bloomberg_prices` accepts xbbg-style wide MultiIndex data and pdblp-style long data. The Bloomberg client remains optional because it requires a configured Bloomberg environment.
